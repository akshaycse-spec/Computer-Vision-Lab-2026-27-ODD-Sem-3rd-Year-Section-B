Experiment No. 1    Experiment Name    Implementation of Fundamental Image Processing Operations using Python and OpenCV       Aim    To develop a comprehensive understanding of fundamental image processing techniques by   implementing image acquisition, storage, color space conversion, geometric transformations,   and  image  enhancement  operations  using  Python  and  OpenCV,  thereby  establishin g  the   foundation for advanced computer vision applications.       Purpose    Image processing forms the core of modern computer vision systems used in domains such as   autonomous  driving,  healthcare,  surveillance,  robotics,  and  industrial  automation.  Before   performing advanced tasks like feature extraction, object detection, or imag e classification, it   is essential to understand how digital images are represented, manipulated, and transformed.    In this experiment, students will implement fundamental image processing operations using   Python and OpenCV . They will learn to read, write, convert, resize, and manipulate digital   images while understanding how these operations influence image quality and prepare images   for subsequent computer vision tasks.       Prerequisites    •   Basic knowledge of Python programming    •   Familiarity with digital image representation (pixels, resolution, color channels)    •   Python libraries: OpenCV (cv2), NumPy, and Matplotlib    •   Jupyter Notebook, Google Colab, or Visual Studio Code       Steps    1.   Install and import the required libraries, including OpenCV, NumPy, and Matplotlib.    2.   Load a color image from the local system and display it using both OpenCV and   Matplotlib.    3.   Examine the image properties such as dimensions, resolution, number of channels, and   data type.    4.   Save the image in different file formats (e.g., JPEG and PNG) and compare the outputs.    5.   Convert the image from RGB/BGR to Grayscale, HSV, and LAB color spaces.

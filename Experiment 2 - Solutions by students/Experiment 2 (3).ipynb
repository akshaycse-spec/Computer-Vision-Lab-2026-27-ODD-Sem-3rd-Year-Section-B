Experiment No. 2    Experiment Name    Contrast Enhancement and Histogram - Based Image Processing using Python and OpenCV       Aim    To analyze and implement contrast enhancement techniques using histogram analysis and   histogram equalization in Python and OpenCV for improving image quality and enhancing   visual information in low - contrast images.       Purpose    Image contrast plays a significant role in the performance of computer vision systems. Poor   illumination,  sensor  limitations,  or  environmental  conditions  often  result  in  low - contrast   images,  making  feature  extraction  and  object  recognition  difficult.  Histo gram - based   enhancement techniques improve the distribution of pixel intensities, thereby enhancing image   visibility and facilitating subsequent image analysis tasks.    In  this  experiment,  students  will  implement  contrast  stretching,  histogram  analysis,  and   histogram equalization using Python and OpenCV. The effectiveness of these enhancement   techniques will be evaluated on real - world images such as medical images, satell ite images, or   low - light photographs.       Prerequisites    •   Basic knowledge of digital image processing    •   Familiarity with Python programming    •   Python libraries: OpenCV (cv2), NumPy, Matplotlib    •   Understanding of grayscale images and pixel intensity distributions    •   Jupyter Notebook, Google Colab, or Visual Studio Code       Steps    1.   Import the required libraries, including OpenCV, NumPy, and Matplotlib.    2.   Load  a  low - contrast  grayscale  image  or convert  a  color  image  into  grayscale  for   processing.    3.   Display the original image and generate its histogram to analyze the distribution of   pixel intensity values.    4.   Perform contrast stretching to improve the dynamic range of pixel intensities.

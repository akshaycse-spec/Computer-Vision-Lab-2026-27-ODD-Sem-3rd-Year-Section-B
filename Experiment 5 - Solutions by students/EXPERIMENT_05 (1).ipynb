```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import hog
from skimage import exposure
```

```python
img = cv2.imread("imgg.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Image loaded successfully")
print("Image Size:", img.shape)
```

```python
plt.figure(figsize=(8,6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
sift = cv2.SIFT_create()

keypoints, descriptors = sift.detectAndCompute(gray, None)

print("Number of SIFT Keypoints:", len(keypoints))
print("Descriptor Shape:", descriptors.shape)
```

```python
sift_img = cv2.drawKeypoints(
    img,
    keypoints,
    None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)

plt.figure(figsize=(8,6))
plt.imshow(cv2.cvtColor(sift_img, cv2.COLOR_BGR2RGB))
plt.title("SIFT Keypoints")
plt.axis("off")
plt.show()
```

```python
features, hog_image = hog(
    gray,
    orientations=9,
    pixels_per_cell=(8,8),
    cells_per_block=(2,2),
    visualize=True
)

print("HOG Feature Length:", len(features))
```

```python
hog_image = exposure.rescale_intensity(
    hog_image,
    in_range=(0,10)
)

plt.figure(figsize=(8,6))
plt.imshow(hog_image, cmap="gray")
plt.title("HOG Features")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(sift_img, cv2.COLOR_BGR2RGB))
plt.title("SIFT Keypoints")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(hog_image, cmap="gray")
plt.title("HOG Features")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("========== RESULT ==========")
print("SIFT Keypoints:", len(keypoints))
print("SIFT Descriptor Shape:", descriptors.shape)
print("HOG Feature Length:", len(features))

print("\nObservation:")
print("SIFT detects distinctive keypoints in the image.")
print("HOG extracts shape and edge information using image gradients.")
print("SIFT is useful for image matching and recognition.")
print("HOG is useful for object detection and shape analysis.")
```
